### **1.1. Selección de tecnologías**
---

- ¿Cuáles son las tecnologías seleccionadas para el proyecto?

- ¿Cómo se alinean las tecnologías seleccionadas con los objetivos del proyecto?

* **¿Cuáles tecnologías se seleccionaron?**
  - Spark se eligió porque el dataset en Parquet es demasiado grande para cargarlo completo con pandas en Colab, mientras que Spark permite leer, filtrar y reducir el volumen de datos antes de moverlos a otras capas y la opción de trabajar con módulos de ML.
  - MongoDB Atlas se eligió porque el proyecto plantea un componente NoSQL tipo Data Lake, y MongoDB permite guardar documentos flexibles con solo las métricas relevantes para el caso de uso.
  - PostgreSQL se eligió porque encaja muy bien con la parte analítica del proyecto: consultas SQL, agregaciones, métricas de negocio y una tabla estructurada para alimentar el modelo.
  - Colab se eligió por disponibilidad gratuita, capacidad de cómputo y facilidad para trabajo en equipo
* **¿Por qué se eligieron en lugar de otras alternativas?**
  - Frente a pandas, Spark es más adecuado para archivos grandes y evita el cuello de botella de RAM que apareció inicialmente.
  - Frente a Cassandra, MongoDB Atlas consideramos es más fácil de desplegar y más amigable para documentos JSON y prototipos rápidos.
  - frente a otras herramientas de SQL, PostgreSQL se encuentran mayor cantidad de servidores gratuitos con aceptables capacidades de almacenamiento.
* **¿Cómo se alinean con los objetivos del proyecto?**  
 Con Spark buscamos cubrir la etapa de procesamiento inicial del Parquet grande.MongoDB cubre la facilidad de almacenamiento de datos  relevantes (semietructurados) resultado del preprocesamiento, simulando un Data Lake.PostgreSQL nos ayudaría en la fase analítica estructurada donde se calculan métricas, se preparan subconjuntos limpios y se generan consultas útiles para el modelado de anomalías sobre variables agregadas como volumen de peticiones, errores y latencia.
* **Cuáles son las ventajas clave de cada tecnología?**
  - Spark: procesa Parquet grande, reduce uso de RAM y permite filtrar o muestrear antes de almacenar.
  - MongoDB Atlas: esquema flexible, fácil de conectar desde Python y útil para guardar solo la parte relevante del dataset.
  - PostgreSQL: consultas SQL claras, buen soporte para agregaciones y preparación de datos tabulares para ML.
* **¿Existen desventajas o limitaciones que deben ser abordadas?**
   - Spark esta sujeto a los recursos limitados del entorno gratuito.
   - MongoDB Atlas Free Tier tiene restricciones de almacenamiento y rendimiento, por lo que no conviene cargar todo el dataset sino solo un subconjunto seleccionado.
   - PostgreSQL tambien presenta  restricciones de almacenamiento en servidores gratuitos.

```python
import time
from pymongo import MongoClient

try:
    client = MongoClient('mongodb://localhost:27017/')
    db = client['mi_base_de_datos'] # Tu base de datos
    collection = db['mi_coleccion'] # Tu colección

    # Insertar un documento de prueba
    test_document = {"mensaje": "¡MongoDB funciona!", "timestamp": time.time()}
    collection.insert_one(test_document)
    print("✅ Documento de prueba insertado con éxito.")

    # Recuperar y mostrar el documento
    print("Documentos en la colección:")
    for doc in collection.find():
        print(doc)

except Exception as e:
    print(f"❌ Error al conectar o interactuar con MongoDB: {e}")
```
