### **1.1. Trasfondo del Negocio**
---

**¿Quiénes son los beneficiarios del proyecto? ¿En qué dominio se encuentran?**

Los beneficiarios directos son los equipos de **operaciones tecnológicas (DevOps)**, **seguridad informática (SOC)**, **administradores de sistemas** y **gerentes de TI** de empresas que poseen infraestructura web (comercio electrónico, banca online, medios digitales, startups, etc.). Indirectamente, se benefician los usuarios finales que experimentan un servicio más estable y seguro, así como los dueños del negocio que evitan pérdidas económicas por caídas o ataques.

El dominio es **monitoreo y observabilidad de sistemas**, un área crítica dentro de la gestión de infraestructura tecnológica.

**¿Cuáles son los principales desafíos o problemas de la situación que motivan la necesidad de realizar un proyecto de análisis de datos?**

Las empresas generan **enormes volúmenes de logs** (decenas de gigabytes diarios) provenientes de servidores web (Apache, Nginx, IIS), balanceadores de carga, firewalls y aplicaciones. Los principales problemas son:

1. **Volumen abrumador**: Los logs son tan grandes que no pueden ser revisados manualmente ni procesados con herramientas tradicionales (Excel, scripts simples).
2. **Variedad de formatos**: Los logs pueden estar en formatos semiestructurados (líneas de texto, JSON) o no estructurados, dificultando su análisis unificado.
3. **Velocidad de generación**: Los registros se producen en tiempo real (miles por segundo). Detectar anomalías (caídas, ataques DDoS, errores masivos) requiere procesamiento casi en tiempo real.
4. **Alta tasa de falsos positivos**: Herramientas tradicionales de monitoreo (como umbrales estáticos) generan muchas alarmas innecesarias. Se necesita un modelo inteligente que aprenda el comportamiento normal del sistema.
5. **Almacenamiento costoso**: Guardar logs crudos por largos periodos es caro. Se requiere una estrategia de Data Lake (crudo) + Data Warehouse (agregados) para balancear costo y consultabilidad.

Este proyecto resuelve la necesidad de **automatizar la detección de anomalías** en los logs, aprovechando tecnologías Big Data y Machine Learning.

### **1.2. Alcance**
---

**¿Qué solución se desea implementar?**

Se desea implementar una **plataforma escalable de ingesta, almacenamiento y análisis de logs de servidores web** que:
- Ingesta líneas de log desde archivos o flujos en tiempo real.
- Almacena logs crudos en una base de datos NoSQL (MongoDB) actuando como **Data Lake**.
- Procesa y transforma los logs a un formato estructurado, cargándolos en una base de datos relacional (PostgreSQL) como **Data Warehouse**.
- Calcula métricas agregadas (peticiones por minuto, códigos de error, latencia promedio) mediante consultas SQL complejas.
- Entrena un modelo de detección de anomalías (Isolation Forest, reglas estadísticas o clasificador) sobre los datos históricos.
- Evalúa en tiempo real (o casi real) las nuevas peticiones y genera alertas cuando se detecta un comportamiento anómalo.

**¿Cuáles son los límites y fronteras del proyecto? ¿Qué aspectos se incluyen y cuáles se excluyen?**

**Incluye:**
- Procesamiento batch de logs históricos (archivos .log, .txt).
- Almacenamiento dual: MongoDB para logs crudos, PostgreSQL para datos agregados.
- Extracción de características: timestamp de la ventana, volumen de peticiones, conteo de respuestas 2xx/4xx/5xx, métricas de latencia, etc.
- Consultas SQL para análisis exploratorio (frecuencia de errores, endpoints más llamados, etc.).
- Modelo de detección de anomalías aplicado a datos agregados por minuto (volumen de peticiones, tasa de errores 4xx/5xx, etc.).
- Generación de alertas (archivo de salida o registro en tabla) cuando se supera un umbral dinámico.

**Excluye:**
- Procesamiento en streaming puro (Kafka, Spark Streaming) – se hará por micro-batches cada minuto.
- Detección de anomalías a nivel de cada petición individual (por razones de rendimiento).
- Interfaz gráfica o dashboard interactivo (se entrega un notebook con visualizaciones básicas).
- Análisis de logs de aplicaciones diferentes a servidores web (ej. logs de base de datos, firewalls).
- Integración con sistemas de alerta reales (Slack, correo, PagerDuty) – solo se simula la salida de alertas.

**¿De qué forma el cliente o beneficiario utilizará el producto del proyecto?**

El equipo de operaciones podrá:
- Ejecutar el pipeline de procesamiento sobre logs de un día o varios días.
- Consultar directamente en PostgreSQL las métricas de salud del sistema mediante SQL (ej. `SELECT hora, COUNT(*) FROM requests WHERE status >= 500 GROUP BY hora`).
- Visualizar en el notebook las anomalías detectadas (gráficos de series de tiempo con puntos marcados).
- Recibir un archivo CSV o tabla con las marcas de tiempo y el tipo de anomalía encontrada.
- Utilizar este sistema como base para implementar un monitor automático que ejecute el modelo periódicamente.

En el futuro, el beneficiario podría extender el proyecto para conectarlo a herramientas de alerta real.

## **2. Definición del diseño y/o arquitectura de la solución**
---

En esta sección debe abordar de manera integral la definición del diseño y/o arquitectura de la solución.

### **2.1. Descripción**
---

![Arquitectura](https://raw.githubusercontent.com/edmendezjob/BigData/374c695fb6797d8594dd8d357473c14e02a88ea3/arquitectura.png)


**¿Cómo se estructura la solución propuesta?**

La solución se estructura en un **pipeline de datos** que combina almacenamiento NoSQL y relacional, procesamiento ETL y un modelo de Machine Learning. Las etapas son:

1. **Ingesta**: Los logs del servidor web se leen desde archivo histórico  Parquet o se simula un flujo en tiempo real.
2. **Data Lake (MongoDB)**: Se almacenan los logs crudos sin transformación, conservando la fidelidad original. Esto permite re-procesar o re-analizar en el futuro.
3. **ETL (Extracción, Transformación, Carga)**: Un script en Python parsea cada línea de log, extrae campos (IP, fecha, método, URL, status, tamaño, referer, user-agent) y calcula campos derivados (hora redondeada, tipo de error, etc.).
4. **Data Warehouse (PostgreSQL)**: Los datos estructurados se cargan en tablas relacionales normalizadas o en una tabla de hechos (star schema). Se crean índices para consultas rápidas.
5. **Agregación temporal**: Mediante SQL (funciones ventana, GROUP BY) se calculan métricas por minuto: número total de requests, promedio de tamaño, conteo de errores 4xx, conteo de errores 5xx, tasa de cambio, etc.
6. **Detección de anomalías**: Se entrena un modelo no supervisado (Isolation Forest o umbrales estadísticos como media + 3 sigma) sobre las series temporales de métricas. El modelo identifica minutos anómalos.
7. **Alertas**: Los resultados se almacenan en una tabla de alertas y se exportan a un archivo CSV para su revisión.

**¿Cuáles son los componentes clave de la solución?**

- **MongoDB**: Base de datos NoSQL documental, ideal para almacenar logs semiestructurados sin esquema fijo. Soporta alta escalabilidad horizontal (sharding).
- **PostgreSQL**: Base de datos relacional ACID, utilizada para datos agregados y consultas analíticas complejas (SQL). Permite integración con herramientas de BI.
- **Python (pymongo, psycopg2)**: Orquestación del ETL y la comunicación entre bases de datos.
- **Pandas**: Manipulación y agregación de datos intermedia.
- **Scikit-learn**: Implementación de Isolation Forest para detección de anomalías.
- **Matplotlib/Seaborn**: Visualización de series temporales y puntos anómalos.

**¿Existen opciones de diseño alternativas y cuáles son sus ventajas y desventajas?**

| Alternativa | Ventajas | Desventajas |
|-------------|----------|--------------|
| Usar solo PostgreSQL con particionamiento | Menor complejidad, SQL unificado | Mala performance para logs no estructurados, difícil escalar a terabytes |
| Usar solo MongoDB con agregaciones | Flexibilidad de esquema, escalable horizontal | Las agregaciones son menos potentes que SQL para reportes complejos |
| Usar Elasticsearch + Kibana (ELK) | Dashboard visual poderoso, búsqueda full-text | Mayor consumo de recursos, curva de aprendizaje alta, sin modelo ML integrado |
| Usar Spark Streaming | Procesamiento en tiempo real real | Sobre-dimensionado para un proyecto académico, requiere clúster |

Nuestra solución híbrida (MongoDB + PostgreSQL) es la más equilibrada: aprovecha lo mejor de ambos mundos y demuestra conceptos clave del módulo de Big Data (Data Lake vs Data Warehouse).

### **2.2. Restricciones y escalabilidad**
---

**¿Cuáles son los requisitos funcionales y no funcionales de la solución?**

**Funcionales:**
- El sistema debe poder procesar archivos de log de tamaño ≥ 1 GB (al menos 1 millón de líneas).
- Debe extraer correctamente los campos estándar de logs en formato Common Log Format (CLF) o Combined Log Format.
- Debe almacenar los logs crudos en MongoDB, sin pérdida de información.
- Debe estructurar y cargar los datos en PostgreSQL, con al menos dos tablas: una de hechos (requests) y una de dimensiones (opcional).
- Debe calcular métricas agregadas por minuto y guardarlas en una tabla de series temporales.
- Debe entrenar un modelo de detección de anomalías (al menos un algoritmo) y aplicarlo a los datos de prueba.
- Debe generar un listado de timestamps anómalos y el tipo de anomalía (pico de errores, caída de tráfico, etc.).

**No funcionales:**
- **Rendimiento**: El ETL completo para 1 GB de logs no debe superar los 30 minutos en un equipo estándar.
- **Escalabilidad**: La arquitectura debe ser claramente escalable horizontalmente (añadir más nodos de MongoDB y PostgreSQL con réplicas).
- **Tolerancia a fallos**: El pipeline debe poder reanudarse desde el punto de fallo (por ejemplo, procesar solo logs nuevos).
- **Mantenibilidad**: El código debe estar comentado y seguir buenas prácticas (funciones modulares, control de errores).
- **Seguridad**: No se manejan datos sensibles, pero se simulará el enmascaramiento de IPs si se usan logs reales.

**¿Existen restricciones tecnológicas, presupuestarias o de tiempo?**

- **Tecnológicas**: El equipo utilizará computadores personales con recursos limitados (8-16 GB RAM). Por ello, se trabajará con logs de tamaño reducido (hasta 2 GB) o se muestrearán datasets públicos. No se podrá implementar un clúster distribuido de producción.
- **Presupuestarias**: El proyecto no requiere inversión económica, pues todas las herramientas son de código abierto (MongoDB Community Edition, PostgreSQL, Python). Se usará almacenamiento local.
- **Tiempo**: El desarrollo debe completarse en 5 semanas, priorizando la funcionalidad central sobre optimizaciones avanzadas.

**¿Cómo se abordará la escalabilidad de la solución a medida que crezcan los datos o las necesidades del negocio?**

- **Escalabilidad horizontal de MongoDB**: Para manejar terabytes de logs, se configuraría un clúster con sharding basado en la clave `timestamp` o `server_id`, distribuyendo los datos entre múltiples réplicas.
- **Escalabilidad de PostgreSQL**: Se utilizaría particionamiento por rango de fechas (tablas heredadas o particiones nativas) y, si es necesario, réplicas de solo lectura para consultas analíticas.
- **Procesamiento paralelo**: El ETL se podría paralelizar dividiendo el archivo de log en fragmentos procesados concurrentemente (usando multiprocessing en Python o Spark).
- **Incremento de velocidad**: Para necesidades casi reales, se sustituiría el procesamiento batch por un stream con Kafka y consumidores en tiempo real, almacenando en MongoDB y actualizando agregaciones en PostgreSQL mediante ventanas deslizantes.
- **Adición de nuevas fuentes**: La arquitectura permite agregar nuevos tipos de logs (por ejemplo, logs de balanceador) simplemente añadiendo campos en MongoDB y adaptando el ETL.

**¿Cómo se integrará la nueva solución con los sistemas existentes en la organización?**

La solución se concibe como un **prototipo independiente** que puede ser adaptado a la infraestructura existente de la siguiente manera:
- Los servidores web ya generan logs en rutas conocidas. El pipeline puede leer esos archivos directamente (con permisos adecuados).
- El equipo de operaciones puede conservar sus herramientas de monitoreo actuales y usar este sistema como un **módulo complementario** de detección inteligente.
- Las alertas generadas se pueden exportar en formato JSON o CSV para ser ingeridas por sistemas de ticket (Jira, ServiceNow) o canales de comunicación (Slack mediante webhooks).
- Las tablas de PostgreSQL pueden ser consultadas por herramientas de BI existentes (Tableau, Power BI, Metabase) sin modificar su configuración.
